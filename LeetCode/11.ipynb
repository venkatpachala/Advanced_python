11. Container With Most Water

1. BruteForce 
Time O(n²)
Space O(1)

```python
def max_area(height:list[int])->int:
    max_area=0
    for i in range(len(height)):
        for j in range(i+1,len(height)):
            width=j-i
            water_height=min(height[i],height[j])
            area=width*water_height

            max_area=max(max_area,area)
    return max_area

max_area(height = [1,8,6,2,5,4,8,3,7])
```

2. Optimal (Two-pointer) -> Time - O(n) Space - O(1)

```python
def max_area_optimal(height:list[int])->int:
    max_area=0
    left=0
    right=len(height)-1

    while left<right:
        area=(right-left)*(min(height[left],height[right]))
        max_area=max(max_area,area)

        if height[left]<height[right]:
            left+=1
        else:
            right-=1

    return max_area

max_area_optimal(height = [1,8,6,2,5,4,8,3,7])
max_area_optimal(height=[3,2,5,8,6,0])
max_area_optimal(height=[3,6,7])
```
