974. Subarray Sums Divisible by K

Prefix Sum + Hash Map + Modular Arithmetic

1. Brute force
-> Time Complexity - O(n2)
-> Space O(1) 

```python
def subarray_sum(nums:list[int],k:int)->int:
    count=0
    for i in range(len(nums)):
        total=0
        for j in range(i,len(nums)):
            total+=nums[j]
            if total%k==0:
                count+=1
    return count

subarray_sum(nums = [4,5,0,-2,-3,1], k = 5)
```

2. Optimized — Prefix Sum + Hash Map

Time - O(n)
Space - O(min(n,k))

```python
def subarray_sumDiv(nums:list[int],k:int)->int:
    freq={0:1}
    prefix_sum=0
    count=0

    for num in nums:
        prefix_sum+=num
        remainder=prefix_sum%k
        count+=freq.get(remainder,0)
        freq[remainder]=freq.get(remainder,0)+1

    return count

subarray_sumDiv(nums = [4,5,0,-2,-3,1], k = 5)
```
