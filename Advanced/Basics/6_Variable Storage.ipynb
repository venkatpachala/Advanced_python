```python
a = 5
b = 5

print(id(a))
print(id(b))
print(a is b)
```

Everything in Python is an Objects

Two different or seperate integer objects are created

```python
a = int("10000")
b = int("10000")

print(a is b)
print(id(a))
print(id(b))
```

A Python variable stores a reference (or binding) to an object, not the object itself. Assignment binds a name to an object. Multiple variables can refer to the same object, and rebinding a variable changes only the reference, not the original object. This reference model is fundamental to Python's handling of objects, mutability, and function arguments.
