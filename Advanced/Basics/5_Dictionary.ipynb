Dictionary(dict) is an ordered, mutable mapping of unique hashable keys to values.

```python
person={
    'name': 'Venkat',
    'age': 22,
    'role': 'AI Engineer'
}
```

```python
print(person['name'])
```

Dictionary is implemented using a hash table (Internally)

We use them as Key & Value pair

```python
data= {
    'initial': 34,
    1: 345,
    2: 56,
    3: 56,
    3: 776
}
print(data)
```

```python
print(data['initial'])
print(data.get('initial'))
print(data.get(2))
print(data.get(3))
print(type(data))
```

```python
data=dict(name='venkat', age=23)
print(data)
```

```python
data={
    'key': 1,
    'value':23
}
print(data)
```

```python
data={1: 'venkat'}
print(type(data))
```

```python
data={1}
print(type(data))
```

```python
data= {
    'initial': 34,
    1: 345,
    2: 56,
    3: 56,
    3: 776
}
print(data)
del data[2]
print(data)
```

Updating Values

```python
data[1]=909
print(data)
```
