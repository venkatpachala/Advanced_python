Immutatbiliy of Strings :
Strings in Python are immutable, which means once a string object is created, its content cannot be changed.

Why Immutable ?

1. Memory Optimisation:
2. Faster Hashing (Dict keys) shld not be changed
3. Thread safety
4. Performance (Caching)
