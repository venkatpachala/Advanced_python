List is a built in Datastructure 

-> It is used for storing the ordered collection

```python
mylist=['banana','apple','orange','grape']
print(mylist[1])
```

List is a:
1. Mutable 
2. Ordered
3. Dynamically sized


How List is Mutable ?


```python
mylist[1] = 'kiwi'
print(mylist)
```

```python
mylist.append('kiwi')
print(mylist)
```

```python
items=[10,'hello',3.14,True]
print(items)
```

```python
containers=[mylist,items,[1,2,3]]
print(containers)
```

The list primarily stores references/pointers to python objects

```python
python_list = [10, 20, 30]

print("Python List:", python_list)

for index, value in enumerate(python_list):
    print(f"Index {index}: reference -> {value}, object id: {id(value)}")
```

```python
python_list.append(100)
print(python_list)
print(python_list[1:4])
print(python_list[::-1])
```

```python
mylist2=list()
print(type(mylist2))
```
