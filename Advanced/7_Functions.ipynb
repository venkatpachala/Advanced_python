```python

def analyze_student(name: str, marks: list[int]) -> dict:
    total = sum(marks)

    if marks:
        average = total / len(marks)
        highest = max(marks)
        lowest = min(marks)
    else:
        average = 0.0
        highest = None
        lowest = None

    passed = bool(marks) and average >= 40

    result = {
        "name": name,
        "total": total,
        "average": average,
        "highest": highest,
        "lowest": lowest,
        "passed": passed
    }

    return result


print(analyze_student("Venkat", [85, 70, 95]))
print(analyze_student("Test", []))
```

Inside the function, args is a tuple of the supplied positional arguments.

```python
def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3))
print(add_numbers(10, 20, 30, 40))
```

```python
def create_student(**kwargs):
    return kwargs

student = create_student(
    name="Venkat",
    age=22,
    course="Python"
)

print(student)
```
