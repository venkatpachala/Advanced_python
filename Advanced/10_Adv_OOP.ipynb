Python provides four important attribute built-in functions.

```python

class ModelClient:
    def __init__(self, model_name):
        self.model_name = model_name


client = ModelClient("gpt-4o")

# Read dynamically
print(getattr(client, "model_name"))

# Update dynamically
setattr(client, "model_name", "qwen3:8b")

print(client.model_name)

# Check existence
print(hasattr(client, "temperature"))

# Create a new attribute
setattr(client, "temperature", 0.7)

print(client.__dict__)

# Delete an attribute
delattr(client, "temperature")

print(client.__dict__)
```

```python

class Temperature:

    def __get__(self, instance, owner):
        if instance is None:
            return self

        return instance.__dict__.get(
            "_temperature", 0.7
        )

    def __set__(self, instance, value):
        if (
            isinstance(value, bool)
            or not isinstance(value, (int, float))
            or not 0 <= value <= 2
        ):
            raise ValueError(
                "Temperature must be between 0 and 2"
            )

        instance.__dict__["_temperature"] = value


class ModelClient:

    temperature = Temperature()

    def __init__(self, model_name, temperature):
        self.model_name = model_name
        self.temperature = temperature


client = ModelClient("gpt-4o", 0.7)

print(client.temperature)

client.temperature = 1.2

print(client.temperature)
print(client.__dict__)
```
