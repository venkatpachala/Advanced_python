Collections :

These are Specialised datastructures for common containers patterns

```python
from collections import (
    Counter,
    defaultdict,
    deque,
    namedtuple,
    OrderedDict,
    ChainMap
)
```

# 1. Counter

It is a specialised dictionary

```python
a='bananana'
my_counter = Counter(a)
print(my_counter)
print(my_counter.most_common(1))  # Most common character
print(type(my_counter))  # <class 'collections.Counter'>
print(list(my_counter.elements()))  # Count of 'a'

```

### Examples

```python
text=str(input("Enter a string: "))

frequency=Counter(text)
print(frequency)
```

```python
text = "bananana"
frequency={}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print(frequency)
```

```python
numbers = [1, 1, 1, 2, 2, 3]
count = Counter(numbers)
print(count)
```

```python
words = ["AI", "ML", "AI", "Python", "AI", "Python"]
count = Counter(words)
print(count)
print(count["AI"])  # Most common word
count.update(["AI", "Python"]) # Increments them not update
print(count)
print(count["AI"]) 
print(count["Python"])
```

# 2. Typedtuple

```python
point=namedtuple('point','x,y')
p1=point(10,20)
print(p1.x)
print(p1.y)
print(p1._asdict())  # Convert namedtuple to dictionary
print(p1._fields)  # Get the field names of the namedtuple"
print(p1._replace(x=30))  # Replace the value of x in the namedtuple
print(type(p1))  # <class '__main__.point'>
```

# 3. OrderedDict

```python
ordered_dict = OrderedDict()
ordered_dict['a'] = 1
ordered_dict['b'] = 2
ordered_dict['c'] = 3
print(ordered_dict)  # OrderedDict([('a', 1), ('b', 2), ('c', 3)])

print(ordered_dict.keys())  # odict_keys(['a', 'b', 'c'])
print(ordered_dict.values())  # odict_values([1, 2, 3])

print(ordered_dict.items())  # odict_items([('a', 1), ('b', 2), ('c', 3)])
print(type(ordered_dict))  # <class 'collections.OrderedDict'>
```

# 4. Defaultdict

```python
d=defaultdict(int)
d['a'] += 1
d['b'] += 2
print(d['a'])  # 1
print(d['b'])  # 2
print(d['c'])  # 0 (default value for int)
print(type(d))  # <class 'collections.defaultdict'>
```

# 5. Deque

```python
d=deque()
d.append(1)
d.append(2)
d.appendleft(0)
print(d)  # deque([0, 1, 2])
# print(d.pop())  # 2
# print(d.popleft())  # 0
# print(d)  # deque([1])
# print(type(d))  # <class 'collections.deque'>
print(d.extend([3, 4, 5]))  # Add multiple elements to the right
```
