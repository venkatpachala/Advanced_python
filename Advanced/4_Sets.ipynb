Set Mathematical Operations


````python
python_devs = {"Venkat", "Sam", "Alex"}

ml_devs = {"Venkat", "John", "Alex"}
````

````python
print(python_devs.union(ml_devs))
python_devs & ml_devs
````

````python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)  # Union
print(a & b)  # Intersection
print(a - b)  # Difference
print(a ^ b)  # Symmetric difference
````
