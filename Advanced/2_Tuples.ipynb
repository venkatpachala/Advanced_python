Tuple is Immutable, Fixed Size, Ordered sequence of Reference of Python Objects

```python
t = ([1, 2], [3, 4])
print(t)

# Change the list inside the tuple
t[0][0] = 10
print(t)

t[0].append(5)
print(t)

print(type(t))
```

```python
mytuple=("Max" ,2, "Boston")
print(mytuple)
```

```python
x=10,20,30
print(x)
print(type(x))
```

Tuple Unpacking

```python
user = ("Venkat", 22, "AI Engineer")

name, age, role = user
print(name)
print(age)
print(role)
```

Tuples are Hashable so works like dictionary Keys

```python
point = (10, 20)

hash(point)
```
