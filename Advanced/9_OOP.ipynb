```python
class AIAgent:

    provider = "OpenAI"

    def __init__(self, name, model):
        self.name = name
        self.model = model
        self.requests = 0

    # Instance method
    def respond(self, query):
        if not self.validate_query(query):
            raise ValueError("Invalid query")

        self.requests += 1
        return f"{self.name}: {query}"

    # Class method
    @classmethod
    def from_default(cls, name):
        return cls(name, "gpt-4o")

    # Static method
    @staticmethod
    def validate_query(query):
        return isinstance(query, str) and bool(query.strip())


agent1 = AIAgent.from_default("SupportAgent")

agent2 = AIAgent(
    "CodeReviewAgent",
    "qwen2.5-coder"
)

print(agent1.respond("Where is my order?"))
print(agent2.respond("Review this PR"))

print(agent1.requests)
print(agent2.requests)

print(AIAgent.validate_query("Hello"))
```

```python

class ModelClient:

    provider = "gptsol"

    def __init__(self, model_name, temperature):
        if not self.validate_temperature(temperature):
            raise ValueError("Temperature must be between 0 and 2")

        self.model_name = model_name
        self.temperature = temperature
        self.requests = 0

    # Instance method
    def generate(self, prompt):
        self.requests += 1

        return (
            f"{self.model_name} is responding "
            f"to your prompt: {prompt}"
        )

    # Class method
    @classmethod
    def from_default(cls):
        return cls("gpt-4o", 0.7)

    # Static method
    @staticmethod
    def validate_temperature(value):
        return (
            isinstance(value, (int, float))
            and not isinstance(value, bool)
            and 0 <= value <= 2
        )


agent1 = ModelClient.from_default()

print(agent1.model_name)
print(agent1.temperature)

print(agent1.generate("Explain RAG"))
print(agent1.generate("Explain LangGraph"))

print("Requests:", agent1.requests)

print(ModelClient.validate_temperature(0.5))
print(ModelClient.validate_temperature(-1))
print(ModelClient.validate_temperature(True))
```

Polymorphism — One interface, different implementations

Method Resolution Order (MRO)

```python

class ModelClient:

    def __init__(self, model_name):
        self.model_name = model_name
        self.requests = 0

    def generate(self, prompt):
        self.requests += 1
        return f"{self.model_name}: {prompt}"


class OpenAIClient(ModelClient):

    def generate(self, prompt):
        response = super().generate(prompt)
        return f"[OpenAI] {response}"


class OllamaClient(ModelClient):

    def generate(self, prompt):
        response = super().generate(prompt)
        return f"[Ollama] {response}"


def run_agent(client, prompt):
    return client.generate(prompt)


openai = OpenAIClient("gpt-4o")
ollama = OllamaClient("qwen2.5-coder:7b")

print(run_agent(openai, "Explain RAG"))
print(run_agent(ollama, "Review this PR"))

print(openai.requests)
print(ollama.requests)
```

```python

from abc import ABC, abstractmethod


class ModelClient(ABC):

    @abstractmethod
    def generate(self, prompt):
        pass


class OpenAIClient(ModelClient):

    def generate(self, prompt):
        return f"OpenAI: {prompt}"


class OllamaClient(ModelClient):

    def generate(self, prompt):
        return f"Ollama: {prompt}"


openai = OpenAIClient()
print(openai.generate("Explain RAG"))
```

Protocol - can describe the methods an object must provide without requiring inheritance

This is called structural typing

```python
from typing import Protocol


class ModelProtocol(Protocol):

    def generate(self, prompt: str) -> str:
        ...


class LocalModel:

    def generate(self, prompt: str) -> str:
        return f"Local model: {prompt}"


def run_agent(model: ModelProtocol, prompt: str) -> str:
    return model.generate(prompt)


model = LocalModel()
print(run_agent(model, "Hello"))
```

Protocol + Composition + Dependency Injection

```python

from typing import Protocol


class RetrieverProtocol(Protocol):

    def search(self, query: str) -> list[str]:
        ...


class VectorRetriever:

    def search(self, query: str) -> list[str]:
        return [f"Vector result for {query}"]


class KeywordRetriever:

    def search(self, query: str) -> list[str]:
        return [f"Keyword result for {query}"]


class RAGAgent:

    def __init__(self, retriever: RetrieverProtocol):
        self.retriever = retriever

    def answer(self, query: str) -> list[str]:
        return self.retriever.search(query)


agent1 = RAGAgent(VectorRetriever())
agent2 = RAGAgent(KeywordRetriever())

print(agent1.answer("What is RAG?"))
print(agent2.answer("What is RAG?"))
```

@property - Properties

```python

class ModelClient:

    def __init__(self, temperature):
        self.temperature = temperature

    @property
    def temperature(self):
        return self._temperature

    @temperature.setter
    def temperature(self, value):
        if (
            isinstance(value, bool)
            or not isinstance(value, (int, float))
            or not 0 <= value <= 2
        ):
            raise ValueError(
                "Temperature must be between 0 and 2"
            )

        self._temperature = value


client = ModelClient(0.7)

print(client.temperature)

client.temperature = 1.2

print(client.temperature)
```

```python

from dataclasses import dataclass, field


@dataclass
class AgentConfig:
    name: str
    model: str
    temperature: float = 0.7


class AIAgent:

    def __init__(self, config: AgentConfig):
        self.config = config
        self._requests = 0
        self._history = []

    @property
    def requests(self):
        return self._requests

    def generate(self, prompt):
        self._requests += 1

        response = (
            f"{self.config.model} responding to {prompt}"
        )

        self._history.append({
            "prompt": prompt,
            "response": response,
        })

        return response

    def __call__(self, prompt):
        return self.generate(prompt)

    def __len__(self):
        return len(self._history)

    def __repr__(self):
        return (
            f"AIAgent(name={self.config.name!r}, "
            f"model={self.config.model!r})"
        )


config = AgentConfig(
    name="SupportAgent",
    model="gpt-4o"
)

agent = AIAgent(config)

print(agent("Where is my order?"))
print(agent("Explain RAG"))

print(agent.requests)
print(len(agent))
print(repr(agent))
```

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        return "Some generic sound"

class Dog(Animal):
    def speak(self):           # Override
        return "Woof!"

class Cat(Animal):
    def speak(self):           # Override
        return "Meow!"

class Duck(Animal):
    def speak(self):
        return "Quack!"

# Polymorphic function
def make_animal_speak(animal):
    print(animal.speak())

# Same function, different behavior
make_animal_speak(Dog('buddy'))    # Woof!
make_animal_speak(Cat('kitty'))    # Meow!
make_animal_speak(Duck('quk'))   # Quack!
```

```python

class BankAccount:
    def __init__(self, owner: str, balance: float = 0):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative")

        self.owner = owner
        self._balance = balance

    def deposit(self, amount: float):
        if amount <= 0:
            raise ValueError("Deposit must be positive")

        self._balance += amount
        return self._balance

    def withdraw(self, amount: float):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")

        if amount > self._balance:
            raise ValueError("Insufficient funds")

        self._balance -= amount
        return self._balance

    @property
    def balance(self):
        return self._balance


account = BankAccount("Venkat", 1000)

account.deposit(500)
account.withdraw(300)

print(account.balance)  # 1200
```

```python

from dataclasses import dataclass

@dataclass
class Invoice:
    invoice_id: str
    customer: str
    amount: float
    paid: bool = False

    def __post_init__(self):
        if self.amount < 0:
            raise ValueError("Amount cannot be negative")

    def mark_paid(self):
        self.paid = True


invoice = Invoice("INV001", "Amazon", 15000)

print(invoice)
invoice.mark_paid()
print(invoice.paid)
```

```python

import time
from functools import wraps

def timer(func):

    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()

        try:
            return func(*args, **kwargs)
        finally:
            end = time.perf_counter()
            print(f"{func.__name__}: {end - start:.4f}s")

    return wrapper


@timer
def process_data():
    time.sleep(2)
    return "Processing complete"


print(process_data())

```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5)) 
```
