Dictonaries Store Key-Value Mapping

````python
user={
    "name": "venkat",
    "age": 22,
    "role": "AI developer"
}

#Other way to create a dictionary
user1=dict(name="venkat", age=22, role="AI developer")
````

````python
user["email"] = "venkat@example.com"
````

````python
print(user)
print(user1)
````

````python
pairs=[("name", "venkat"), ("age", 22), ("role", "AI developer")]
print(type(pairs))
print(pairs)

user2=dict(pairs)
print(type(user2))
print(user2)
````

````python
data = {
    "name": "Venkat",
    "age": 22,
    "score": 9.5,
    "skills": ["Python", "RAG"],
    "active": True,
    "metadata": {
        "level": "junior"
    }
}

print(data)
print(type(data))
````

````python
data1 = {
    "name": "Venkat",
    10: "ten",
    3.14: "pi",
    True: "yes",
    (1, 2): "coordinates"
}
print(data1)
````

Dictionaries are hash tables

```python
"name"
   ↓
hash("name")
   ↓
hash value
   ↓
dictionary table location
   ↓
"name" → "Venkat"
```
