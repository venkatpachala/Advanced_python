# String - Immutable Ordered and Text Representation

````python
Name='Venkat'
print(Name)
````

````python
len(Name)
for i in range(len(Name)):
    print(Name[i])
````

````python
for char in Name:
    print(char)
````

````python
for i, char in enumerate(Name):
    print(i, char)
````

# string[start:stop:step]

````python
print(Name[0])
print(Name[-1])
print(Name[0:3])
print(Name[0:6:2])
print(Name[::-1])
print(Name.upper())
print(Name.lower())
# Strings are immutable in Python, which means you 
# cannot change their content after they are created.
# print(Name[0]='v')  # This will raise an error because strings are immutable
````

````python
Name='Venkat'
encoded_name=Name.encode('utf-8')
print(encoded_name)
print(type(encoded_name))
````

````python
decoded = encoded_name.decode("utf-8")
print(decoded)
print(type(decoded))
````

````python
print(ord('A'))  # Returns the Unicode code point for 'A'
print(chr(65))  # Returns the character for Unicode code point 65
````

````python
a = "python"
b = "python"
print(a is b)  # True, because both refer to the same string object in memory

print('python' is 'python')  # True, because both refer to the same string object in memory
````

# Strings are immuatable hence Hashable

````python
Name='Venkat'
print(hash(Name))
print(hash('Venkat'))
````

```python
## | Operation | Complexity |
|---|---:|
| `len(s)` | O(1) |
| `s[i]` | O(1) |
| `s[a:b]` | O(k) |
| iterate string | O(n) |
| copy string content | O(n) |
| `char in s` | O(n) worst-case |
| `s1 == s2` | O(n) worst-case |
| `lower()` | O(n) |
| `upper()` | O(n) |
| `split()` | O(n) |
| `join()` | O(total output size) |
| reverse via slicing | O(n) |
```

````python
text = "bananana"
frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
````

````python
from collections import Counter

freq = Counter(text)
print(freq)
````
