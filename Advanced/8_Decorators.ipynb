```python

from functools import wraps

def my_decorator(func):

    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Function started")

        result = func(*args, **kwargs)

        print("Function finished")
        return result

    return wrapper


@my_decorator
def add(a, b):
    return a + b

print(add(10, 20))
```

```python

from functools import wraps

def repeat(times):

    def decorator(func):

        @wraps(func)
        def wrapper(*args, **kwargs):

            for _ in range(times):
                result = func(*args, **kwargs)

            return result

        return wrapper

    return decorator


@repeat(times=3)
def greet():
    print("Hello!")


greet()
```

```python
from functools import wraps

def repeat(times: int):
    if type(times) is not int or times < 1:
        raise ValueError("times must be a positive integer")

    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(times):
                result = func(*args, **kwargs)
            return result

        return wrapper

    return decorator

@repeat(times=3)
def multiply(a, b):
    print(f"Multiplying {a} and {b}")
    return a * b

print(multiply(4, 5))
```

Build an execution timer decorator

```python

import time
from functools import wraps

def timer(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        start = time.perf_counter()

        result = func(*args, **kwargs)

        end = time.perf_counter()

        elapsed = end - start

        print(f"{func.__name__} took {elapsed:.4f}s")

        return result

    return wrapper


@timer
def retrieve_documents(query):
    time.sleep(0.5)
    return ["Document A", "Document B"]


documents = retrieve_documents("What is RAG?")
print(documents)
```

```python

import logging
import time
from functools import wraps

logger = logging.getLogger(__name__)

def timer(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        start = time.perf_counter()

        try:
            return func(*args, **kwargs)

        finally:
            elapsed = time.perf_counter() - start

            logger.info(
                "Function %s completed in %.4f seconds",
                func.__name__,
                elapsed,
            )

    return wrapper

@timer
def retrieve_documents(query):
    time.sleep(0.5)
    return ["Document A", "Document B"]


documents = retrieve_documents("What is RAG?")
print(documents)
```

```python

import asyncio
import time

async def get_order():
    await asyncio.sleep(2)
    return {"order_id": 123, "status": "shipped"}

async def get_customer():
    await asyncio.sleep(1)
    return {"name": "Alex"}

async def get_return_policy():
    await asyncio.sleep(3)
    return {"return_window_days": 30}


async def agent():
    start = time.perf_counter()

    task1 = asyncio.create_task(get_order())
    task2 = asyncio.create_task(get_customer())
    task3 = asyncio.create_task(get_return_policy())

    order, customer, policy = await asyncio.gather(
        task1,
        task2,
        task3
    )

    elapsed = time.perf_counter() - start

    return {
        "order": order,
        "customer": customer,
        "policy": policy,
        "execution_time": elapsed,
    }


result = await agent()
print(result)
```

### asyncio.TaskGroup 

A TaskGroup lets you create a group of related asynchronous tasks and manage their lifetimes together.

```python
import asyncio
async def agent():
    async with asyncio.TaskGroup() as tg:
        order_task= tg.create_task(get_order())
        customer_task= tg.create_task(get_customer())
        policy_task= tg.create_task(get_return_policy())

    return {
        "order": order_task.result(),
        "customer": customer_task.result(),
        "policy": policy_task.result(),}

result =await agent()
print(result)
```

```python

import asyncio
import time

async def fetch_tool(name, delay, semaphore):
    async with semaphore:
        print(f"Starting {name}")
        await asyncio.sleep(delay)
        print(f"Finished {name}")
        return {"tool": name, "status": "success"}


async def run_agent():
    start = time.perf_counter()

    semaphore = asyncio.Semaphore(2)

    try:
        async with asyncio.timeout(5):

            async with asyncio.TaskGroup() as tg:
                order = tg.create_task(
                    fetch_tool("order", 2, semaphore)
                )

                customer = tg.create_task(
                    fetch_tool("customer", 1, semaphore)
                )

                policy = tg.create_task(
                    fetch_tool("policy", 3, semaphore)
                )

        return {
            "order": order.result(),
            "customer": customer.result(),
            "policy": policy.result(),
            "elapsed": time.perf_counter() - start,
        }

    except TimeoutError:
        return {"error": "Agent execution timed out"}


result = await run_agent()
print(result)
```
