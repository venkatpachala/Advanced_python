1. Two Sum

```python
nums = [3, 3]
target = 6
def twosum(nums:list[int],target:int):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return i,j

twosum(nums,target)
```

2. Contains Duplicates

```python
s=[1, 2, 3]

def duplicates_contain(s:list[int]):
    for i in range(len(s)):
        for j in range(i+1,len(s)):
            if s[i]==s[j]:
                return True
    return False

duplicates_contain(s)
```

```python
s=[1, 2, 3,1]

def contains_duplicates(s):
    seen=set()

    for i in s:
        if i in seen:
            return True
        seen.add(i)
    return False

contains_duplicates(s)
```

3. Valid Anagram

```python
s = "anagram"
t = "nagaram"

def anagram(s:str,t:str):
    if sorted(s)==sorted(t):
        return True
    else:
        return False

anagram(s,t)
```

6. Group Anagrams

```python
s = ["eat", "tea", "tan", "ate", "nat", "bat"]

def group_anagrams(s):
    groups = {}

    for word in s:
        key = "".join(sorted(word))
        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())

print(group_anagrams(s))
```

7. Top K Frequency Elements

```python
nums = [1, 1, 1, 2, 2, 3]
k = 2

def freq_elements(nums: list[int], k: int):
    result = {}

    for i in range(len(nums)):
        count = 0
        for j in range(len(nums)):
            if nums[i] == nums[j]:
                count += 1
        result[nums[i]] = count

    sorted_result = sorted(
        result,
        key=result.get,
        reverse=True
    )

    return sorted_result[:k]


print(freq_elements(nums, k)) 
```

8. Product of array elements except self

```python
nums = [1, 2, 3, 4]

def product_array(nums:list[int]):
    result=[]
    for i in range(len(nums)):
        product=1
        for j in range(len(nums)):
            if i!=j:
                product*=nums[j]
        result.append(product)
    return result

product_array(nums)
```

9. Majority Element

```python
nums=[2, 2, 1, 1, 1,1,1, 2, 2,2,2]

def majority_ele(nums):
    result={}
    for i in range(len(nums)):
        count=0

        for j in range(len(nums)):
            if nums[i]==nums[j]:
                count+=1
        result[nums[i]]=count

        if count> len(nums)//2:
            return nums[i]


majority_ele(nums)
```

10. Intersection of Two Arrays

```python
nums1 = [1, 2, 2, 1]
nums2 = [2, 2]

def intersection(nums1, nums2):
    result = []

    for i in range(len(nums1)):
        for j in range(len(nums2)):
            if nums1[i] == nums2[j]:
                if nums1[i] not in result:
                    result.append(nums1[i])

    return result


# nums1 = [4, 9, 5]
# nums2 = [9, 4, 9, 8, 4]

print(intersection(nums1, nums2))
```

11. Longest Substrings without repeatinf characters

```python
s = "abcabcbb"
def long_substring(s:str)->int:
    s=list(s)
    max_length = 0
    for i in range(len(s)):
        result=[]
        for j in range(i,len(s)):
            if s[j] not in result:
                result.append(s[j])
            else:
                break
        max_length = max(max_length, len(result))
    return max_length  
print(long_substring(s)) 
```

12. Valid Plaindrome in a string

```python
def is_palindrome(s):
    cleaned = ""

    for char in s:
        if char.isalnum():
            cleaned += char.lower()

    return cleaned == cleaned[::-1]


print(is_palindrome("A man, a plan, a canal: Panama"))
```

```python

s = "babad"

def palindrome_string(s: str):
    result = ""

    for i in range(len(s)):
        for j in range(i + 1, len(s) + 1):

            substring = s[i:j]

            if substring == substring[::-1]:
                if len(substring) > len(result):
                    result = substring

    return result

print(palindrome_string(s))
```

```python
nums=[2,2]
count=0
def distinct_sum(nums:list[int]):
    subarray=[]
    for i in range(len(nums)):
        for j in range(i, len(nums)):
            subarray.append(nums[i:j+1])
        total=0

        for arr in subarray:
            result=[]
            for k in arr:
                if k not in result:
                    result.append(k)
            count=len(result)
            total+=count**2
    return total

print(distinct_sum(nums))
```

```python
def three_sum(nums):
    result = set()
    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):

                if nums[i] + nums[j] + nums[k] == 0:
                    triplet = tuple(sorted([
                        nums[i], nums[j], nums[k]
                    ]))
                    result.add(triplet)

    return [list(t) for t in result]


nums = [-1, 0, 1, 2, -1, -4]
print(three_sum(nums))
# [[-1, -1, 2], [-1, 0, 1]]
# Order may vary
```
