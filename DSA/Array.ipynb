1. Two Sum

```python
nums=[-1,-2,-3,-4,-5]
target=-8

def two_sum(nums,target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return i,j

two_sum(nums,target)
```

```python
def two_sum(nums,target):
    seen={}
    for i in range(len(nums)):
        complement=target-nums[i]

        if complement in seen:
            return [seen[complement],i]

        seen[nums[i]]=i
    return []
two_sum(nums,target)
```

2. Contains Duplicate

```python
nums=[1,2,3,3]
result=[]
def duplicate(nums):
    for i in range(len(nums)):
        if nums[i] in result:
            return True
        else:
            result.append(nums[i])
    return False

duplicate(nums)
```

```python
def contain_duplicate(nums):
    seen=set()

    for num in nums:
        if num in seen:
            return True

        seen.add(num)
    return False

contain_duplicate(nums)
```

3. Contains Duplicate II


```python
nums = [1, 2, 3, 1]
k = 3
result=[]
def contain_duplicate2(nums,k):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]==nums[j] and j-i<=k:
                return True
    return False    

contain_duplicate2(nums,k)        
```

Sliding Window (set)

```python
nums = [1, 2, 3,1]
k = 3

def contain_duplicate22(nums,k):
    seen=set()
    for i in range(len(nums)):
        if nums[i] in seen:
            return True

        seen.add(nums[i])

        if len(seen)>k:
            seen.remove(nums[i-k])
    return False

contain_duplicate22(nums,k)
```

4. Remove Duplicates from Sorted Array

Intro to Two Pointers

```python
nums=[0, 0, 1, 1, 2, 2, 3]

def remove_duplicate(nums):
    write=1
    for read in range(1,len(nums)):
        if nums[read]!=nums[write-1]:
            nums[write]=nums[read]
            write+=1
    return write

k = remove_duplicate(nums)
print(k)         
print(nums[:k])  
```

6. Best Time to Buy and Sell Stock

```python
nums = [7, 1, 5, 3, 6, 4]
def max_profit(nums):
    max=0
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            profit = nums[j] - nums[i]
            if profit > max:
                max=profit
    return max
max_profit(nums)
```

```python
def max_profit(prices):
    min_price = float('inf')
    maximum_profit = 0

    for price in prices:
        min_price = min(min_price, price)

        profit = price - min_price

        maximum_profit = max(maximum_profit, profit)

    return maximum_profit


nums = [7, 1, 5, 3, 6, 4]
```

7. Maximum Subarray (Kadane's Algorithm)

Dynamic Programming

Brute Force

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

def max_subarray(nums):
    max_sum=float('-inf')
    for i in range(len(nums)):
        current_sum=0
        for j in range(i,len(nums)):
            current_sum+=nums[j]
            if current_sum>max_sum:
                max_sum=current_sum
    return max_sum

print(max_subarray(nums))
```

Kadane's Optimisation

```python
def max_subarray(nums):
    current_sum = nums[0]
    max_sum = nums[0]

    for i in range(1, len(nums)):
        current_sum = max(
            nums[i],
            current_sum + nums[i]
        )

        max_sum = max(max_sum, current_sum)

    return max_sum

max_subarray(nums)
```

8. Merge Sorted Arrays

```python
nums1 = [1, 2, 3, 0, 0, 0]
m = 3
nums2 = [2, 5, 6]
n = 3

def merge_array(nums1,m,nums2,n):
    for i in range(n):
        nums1[m+i]=nums2[i]

    nums1.sort()
    return nums1

merge_array(nums1, 3, nums2, 3)
```

3 Pointers (Optimisation)

```python
def merge(nums1, m, nums2, n):
    i = m - 1
    j = n - 1
    k = m + n - 1

    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1

        k -= 1


nums1 = [1, 2, 3, 0, 0, 0]
nums2 = [2, 5, 6]

merge(nums1, 3, nums2, 3)

print(nums1)
# [1, 2, 2, 3, 5, 6]
```

9. Product of Array Except Self.

```python
nums = [1, 2, 3, 4]

def product_array(nums:list[int]):
    result=[]
    for i in range(len(nums)):
        temp=1
        for j in range(len(nums)):
            if i!=j:
                temp*=nums[j]
        result.append(temp)
    return result
product_array(nums)
```

Optimised

```python
def product_except_self(nums):
    n = len(nums)
    result = [1] * n

    # Pass 1: Prefix products
    prefix = 1

    for i in range(n):
        result[i] = prefix
        prefix *= nums[i]

    # Pass 2: Suffix products
    suffix = 1

    for i in range(n - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]

    return result


nums = [1, 2, 3, 4]
print(product_except_self(nums))

# [24, 12, 8, 6]
```

Factorial

```python
nums=5

def factorial(nums:int):
    result=1
    for i in range(1,nums+1):
        result*=i
    return result

factorial(nums)
```

10. Maximum Consecutive Ones

```python
def max_ones(nums):
    max_count=0
    count=0
    for i in range(len(nums)):
        count=0
        for j in range(i,len(nums)):
            if nums[j]==1:
                count+=1
            else:
                break
        max_count = max(max_count, count)
    return max_count

max_ones(nums=[1,1,0,1,1,1])
```

Optimised

```python
def max_consecutive_ones(nums):
    count = 0
    max_count = 0

    for num in nums:
        if num == 1:
            count += 1
        else:
            count = 0

        max_count = max(max_count, count)

    return max_count


nums = [1, 1, 0, 1, 1, 1]

print(max_consecutive_ones(nums))  
```
