1. Valid Anagram

```python
s = "anagram"
t = "nagaram"

def is_anagram(s,t):
    if sorted(s)==sorted(t):
        return True
    else:
        return False
    
is_anagram(s,t)
```

2. First Non repeating character

```python
def first_non_repeating(s):
    for i in range(len(s)):
        count = 0

        for j in range(len(s)):
            if s[i] == s[j]:
                count += 1

        if count == 1:
            return s[i]

    return None


print(first_non_repeating("swiss"))  
```

```python
s="swiss"

def non_rep(s:str):
    freq={}
    for char in s:
        freq[char]=freq.get(char,0)+1

    for char in s:
        if freq[char]==1:
            return char
        
non_rep(s)
```

3. Character Frequency

```python
def char_freq(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1

    return freq


print(char_freq("swiss"))
```

4. Group Anagrams

```python
s = ["eat", "tea", "tan", "ate", "nat", "bat"]

def group_anagrams(s):
    groups = {}

    for word in s:
        key = "".join(sorted(word))
        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())

print(group_anagrams(s))
```

5. Valid Palindrome

```python
def is_palindrome(s):
    cleaned = ""

    for char in s:
        if char.isalnum():
            cleaned += char.lower()

    return cleaned == cleaned[::-1]


s = "A man, a plan, a canal: Panama"
print(is_palindrome(s))
```
