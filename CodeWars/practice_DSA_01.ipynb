```python
def document_stats(text: str) -> dict:
    characters = len(text)

    words = text.split()
    word_count = len(words)

    lowercase_words = [word.lower() for word in words]
    unique_words = len(set(lowercase_words))

    if words:
        longest_word = max(words, key=len)
    else:
        longest_word = None

    result = {
        "characters": characters,
        "words": word_count,
        "unique_words": unique_words,
        "longest_word": longest_word
    }

    return result


text = input("Enter String: ")
print(document_stats(text))
```

```python
def character_freq(text: str) -> dict[str, int]:
    freq = {}

    text = text.lower()

    for char in text:
        if char == " ":
            continue

        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    return freq


print(character_freq("Hello World"))
```

```python

def first_non_repeating(text: str) -> str | None:
    text = text.lower()
    freq = {}

    # Pass 1: Count character frequencies
    for char in text:
        if char == " ":
            continue

        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    # Pass 2: Find the first unique character
    for char in text:
        if char == " ":
            continue

        if freq[char] == 1:
            return char

    return None


print(first_non_repeating("leetcode"))
```

### Two Sum

BruteForce

```python
def two_sum(nums: list[int],target: int)-> list[int]:
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i] + nums[j] == target:
                return [i,j]
    return []

print(two_sum(nums = [2, 8, 8, 1],target = 9))
```

Optimised

```python

def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []


print(two_sum([2, 7, 11, 15], 9))
```

### Anagrams

```python

def group_anagrams(words: list[str]) -> list[list[str]]:
    groups = {}

    for word in words:
        key = tuple(sorted(word))

        if key in groups:
            groups[key].append(word)
        else:
            groups[key] = [word]

    return list(groups.values())


print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

### Valid Paranthesis

```python

def is_valid_parentheses(s: str) -> bool:
    stack = []

    pairs = {
        ")": "(",
        "]": "[",
        "}": "{"
    }

    for char in s:
        if char in "([{":
            stack.append(char)

        elif char in pairs:
            if not stack:
                return False

            if stack[-1] != pairs[char]:
                return False

            stack.pop()

    return len(stack) == 0


print(is_valid_parentheses("{[)}"))
```

### Valid Anagram

```python

def anagram(s: str, t: str) -> bool:
    key = tuple(sorted(s))
    key1 = tuple(sorted(t))

    if key == key1:
        return True
    else:
        return False


print(anagram("anagram", "nagaram"))  # True
print(anagram("rat", "car"))          # False
```

```python

def is_anagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False

    freq = {}

    # Count characters in s
    for char in s:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    # Subtract characters from t
    for char in t:
        if char not in freq:
            return False

        freq[char] -= 1

        if freq[char] < 0:
            return False

    return True


print(is_anagram("nagaram", "anagram"))
```

Best time to buy and sell stock

```python

def max_profit(prices: list[int]) -> int:
    max_profit_value = 0

    for i in range(len(prices)):
        for j in range(i + 1, len(prices)):
            profit = prices[j] - prices[i]

            if profit > max_profit_value:
                max_profit_value = profit

    return max_profit_value


print(max_profit([7, 1, 5, 3, 6, 4]))
```

```python
def max_subarray(nums:list[int])->list[int]:
    current_sum=nums[0]
    max_sum=nums[0]

    for i in range(1,len(nums)):
        num=nums[i]
        current_sum=max(num,current_sum+num)
        max_sum=max(max_sum,current_sum)
    
    return max_sum

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

### Move Zeros

```python
def move_zeros(nums: list[int])-> list[int]:
    result=[]
    zeros=[]
    for i in range(len(nums)):
        if nums[i]!=0:
            result.append(nums[i])
        elif nums[i]==0:
            zeros.append(nums[i])
    result.sort()
    return result+zeros

print(move_zeros(nums = [0, 1, 0, 3, 12]))
```

```python

def move_zeros(nums: list[int]) -> None:
    write = 0

    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write], nums[read] = nums[read], nums[write]
            write += 1


nums = [0, 1, 0, 3, 12]
move_zeros(nums)
print(nums)
# [1, 3, 12, 0, 0]
```

```python

def remove_duplicates(nums: list[int]) -> int:
    if not nums:
        return 0

    write = 1

    for read in range(1, len(nums)):
        if nums[read] != nums[write - 1]:
            nums[write] = nums[read]
            write += 1

    return write


nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = remove_duplicates(nums)

print(k)
print(nums[:k])
```

Two Pointer Method


```python

def is_palindrome(s: str) -> bool:
    left = 0
    right = len(s) - 1

    while left < right:

        # Skip non-alphanumeric on left
        if not s[left].isalnum():
            left += 1
            continue

        # Skip non-alphanumeric on right
        if not s[right].isalnum():
            right -= 1
            continue

        # Compare lowercase characters
        if s[left].lower() != s[right].lower():
            return False

        # Move inward
        left += 1
        right -= 1

    return True


print(is_palindrome("racecar"))  # True
print(is_palindrome("race a car"))  # False
print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome(""))  # True
```

```python

def length_of_longest_substring(s: str) -> int:
    seen = set()
    left = 0
    max_length = 0

    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        current_length = right - left + 1
        max_length = max(max_length, current_length)

    return max_length


print(length_of_longest_substring("abcabcbb"))  # 3
print(length_of_longest_substring("bbbbb"))     # 1
print(length_of_longest_substring("pwwkew"))    # 3
print(length_of_longest_substring("abba"))      # 2
print(length_of_longest_substring(""))          # 0
```
