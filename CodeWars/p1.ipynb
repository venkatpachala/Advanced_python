1. Write Number in Expanded Form


```python
num=int(input("Enter an integer: "))
```

```python
def expanded_form(num):
    result = []
    text = str(num)
    
    for i in range(len(text)):
        digit = text[i]

        if digit != "0":
            zeros = len(text) - i - 1
            ones = digit + "0" * zeros
            result.append(ones)
            
    return " + ".join(result)

print(expanded_form(num))
```

```python
s=str(input("Enter a string: "))
```

```python
def solution(s):
    result = ""

    for char in s:
        if char.isupper():
            result += " "

        result += char

    return result

print(solution(s))
```
