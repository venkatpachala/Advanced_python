```python

nums1 = [1, 3]
nums2 = [2]

def findMedianSortedArrays(nums1, nums2):
    nums = nums1 + nums2
    nums.sort()

    n = len(nums)
    mid = n // 2

    if n % 2 == 1:
        return float(nums[mid])
    else:
        return (nums[mid - 1] + nums[mid]) / 2

print(findMedianSortedArrays(nums1, nums2))
```

5. Longest Palindromic Substring

```python

s = "babad"

def palindrome_string(s: str):
    result = ""

    for i in range(len(s)):
        for j in range(i + 1, len(s) + 1):

            substring = s[i:j]

            if substring == substring[::-1]:
                if len(substring) > len(result):
                    result = substring

    return result

print(palindrome_string(s))
```

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        result = ""

        for i in range(len(s)):
            # Odd-length palindrome
            left = right = i

            while left >= 0 and right < len(s) and s[left] == s[right]:
                if right - left + 1 > len(result):
                    result = s[left:right + 1]

                left -= 1
                right += 1

            # Even-length palindrome
            left, right = i, i + 1

            while left >= 0 and right < len(s) and s[left] == s[right]:
                if right - left + 1 > len(result):
                    result = s[left:right + 1]

                left -= 1
                right += 1

        return result
```

```python
def palindrome(s: str):
    longest = ""

    for i in range(len(s)):
        for j in range(i, len(s)):

            substring = s[i:j+1]

            if substring == substring[::-1]:
                if len(substring) > len(longest):
                    longest = substring

    return longest


print(palindrome("babad"))
print(palindrome("cbbd"))
print(palindrome("racecar"))
```

```python
nums=[22,14,33,7]
def swap(nums:list[int]):
    result=[]
    result=[sum(int(digit) for digit in str(number)) for number in nums]
    return sorted(result)

print(swap(nums))
```

```python
class Solution:
    def minSwaps(nums: list[int]) -> int:

        def digit_sum(num):
            return sum(int(d) for d in str(num))

        target = sorted(nums, key=lambda x: (digit_sum(x), x))

        position = {num: i for i, num in enumerate(nums)}

        swaps = 0

        for i in range(len(nums)):

            if nums[i] != target[i]:

                correct = target[i]
                j = position[correct]

                position[nums[i]] = j
                position[correct] = i

                nums[i], nums[j] = nums[j], nums[i]

                swaps += 1

        return swaps
```
