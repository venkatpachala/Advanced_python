1. Dependency injection without FastAPI

```python
class LLMClient:
    def generate(self, prompt: str):
        return f"Answer to: {prompt}"

class InvoiceAgent:
    def __init__(self):
        self.llm = LLMClient()

    def process(self, question: str):
        return self.llm.generate(question)

agent = InvoiceAgent()
print(agent.process("Explain this invoice"))
```

```python
class InvoiceAgent:
    def __init__(self, llm):
        self.llm = llm

    def process(self, question: str):
        return self.llm.generate(question)


class OllamaClient:
    def generate(self, prompt: str):
        return f"Ollama response: {prompt}"


class FakeClient:
    def generate(self, prompt: str):
        return "Test response"
```

```python
agent = InvoiceAgent(OllamaClient())
print(agent.process("What is the due date?"))
```

```python
test_agent = InvoiceAgent(FakeClient())
print(test_agent.process("What is the due date?"))
```
