### 2. Path and Query parameters

Path parameter

Path parameters identify a resource in the URL.

```python
from fastapi import FastAPI

app = FastAPI()


@app.get("/agents/{agent_id}")
def get_agent(agent_id: int):
    return {"agent_id": agent_id}
```

Path Validation
(the input shld be gt=0 means greater than 0)

```python
from typing import Annotated
from fastapi import Path

@app.get('/agents/{agent_id}')
def get_agent(
    agent_id: Annotated[int,Path(gt=0, description="Unique agent ID")]
):
    return {"agent_id":agent_id}
```

Multiple path parameters

```python
@app.get("/organization/{org_id}/agents/{agents_id}")
def get_org_agent(
    org_id:int,
    agent_id:int
):
    return{
        "organization_id": org_id,
        "agent_id":agent_id
    }
```

Query Parameter

Query parameters generally represent filtering, sorting, pagination, and optional request configuration.

```python
@app.get("/agents")
def list_agents(limits: int=10, offset:int=0
):
    return {
        "limit": limits,
        "offset":offset

    }
```
